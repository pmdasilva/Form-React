```python
import os
from selenium import webdriver
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.by import By
from dotenv import load_dotenv
from selenium import webdriver
from selenium.webdriver.common.by import By

load_dotenv()  # carrega o .env
```

```python
class PortalInfomacoes:
    def __init__(self):
        self.driver = webdriver.Edge()
        self.site = os.getenv("SITEPORTALINFORMACOES")
        self.PASSWORD = os.getenv("PASSWORDPORTAL")
        self.LOGIN = os.getenv("LOGINPOTAL")
        self.input_login = "user_login"
        self.input_password = "user_pass"
        self.submit = "wp-submit"

    def time_wait_window(self):
        self.driver.implicitly_wait(2)

    def open_window(self):
        # This method wiil opening this windown and waiting 2 seconds.
        self.driver.get(self.site)
        self.time_wait_window()

    def select_input_login(self):
        # This method select this advanced button.
        self.driver.find_element(By.ID, self.input_login).click()

    def fill_input_login(self):
        # This method insert the camp email of loginWiki:
        input_email = self.driver.find_element(By.ID, self.input_login)
        input_email.send_keys(self.LOGIN)

    def select_input_password(self):
        # This method select this advanced button.
        self.driver.find_element(By.ID, self.input_password).click()

    def fill_user_password(self):
        # This method insert the camp password of loginWiki:
        input_password = self.driver.find_element(By.ID, self.input_password)
        input_password.send_keys(self.PASSWORD)

    def select_button_submit(self):
        # This method slected this button login.
        self.driver.find_element(By.ID, self.submit).click()
```

```python
acesso = PortalInfomacoes()
acesso.open_window()
acesso.select_input_login()
acesso.fill_input_login()
acesso.select_input_password()
acesso.fill_user_password()
acesso.select_button_submit()
```
